```python
# Data visualization on Python

# matplotlib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Lets create a data and create a line chart

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales = [100, 150, 200, 170, 160, 210, 230, 300, 310, 270, 290, 300]

plt.plot(months, sales) # it makes a line chart
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Monthly Sales')
plt.show()
```

```python
data = pd.read_excel('/content/sample_data/Global_Superstore2.xlsx')
data
```

```python
pd.set_option('display.float_format', '{:.2f}'.format)
data.head()

#Feature Engineering (Dates & Months)
data['Order Date'] = pd.to_datetime(data['Order Date'])

data['order month'] = data['Order Date'].dt.month

#Data Aggregation
agg_data = data.groupby('order month')['Sales'].sum().reset_index()


plt.plot(agg_data['order month'], agg_data['Sales'].astype(int),'o--g')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales')
plt.grid(axis = 'y')
plt.show()
```

```python
# Bar chart to compare categories

# I want to see which category generates highest sales

agg_data = data.groupby('Category')['Sales'].sum().reset_index()
agg_data = agg_data.sort_values(by='Sales', ascending=False)
agg_data

plt.bar(agg_data['Category'], agg_data['Sales'])
plt.xlabel('Category')
plt.ylabel('Sales')
```

```python
agg_data = data.groupby('Category')['Sales'].sum()

agg_data.plot(kind='bar',figsize=(5,5))
plt.xlabel('Category')
plt.ylabel('Sales')
```

```python
# Create two line charts :

'''1. Monthly sales
2. Yearly sales'''
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])


# Monthly
data['order month'] = data['Order Date'].dt.month

agg_data = data.groupby('order month')['Sales'].sum().reset_index()

plt.plot(agg_data['order month'], agg_data['Sales'].astype(int),'o--g')

# Yearly
data['year'] = data['Order Date'].dt.year

agg_data = data.groupby('year')['Sales'].sum().reset_index()

plt.plot(agg_data['year'], agg_data['Sales'].astype(int),'o--g')
```

```python
# Create 4 charts
#1. day wise sales
#2. week wise sales
#3. month wise sales
#4. year wise sales

data
```

```python
data['day'] = data['Order Date'].dt.day
data['week'] = data['Order Date'].dt.isocalendar().week
data['month'] = data['Order Date'].dt.month
data['year'] = data['Order Date'].dt.year

data

time_list = ['day','month','week','year']

plt.figure(figsize=(15,10))

for i in range(len(time_list)):
  agg_data = data.groupby(time_list[i])['Sales'].sum().reset_index()
  plt.subplot(2, 2, i + 1)
  plt.plot(agg_data[time_list[i]], agg_data['Sales'],'o--g')
  plt.xlabel(time_list[i])
  plt.ylabel('Sales')
  plt.title(time_list[i])
```

```python
sub_cat_profit = data.groupby('Sub-Category')['Profit'].sum().sort_values(ascending = False).reset_index()

plt.figure(figsize = (10,5))

bar = plt.barh(sub_cat_profit['Sub-Category'], sub_cat_profit['Profit'])
plt.bar_label(bar, padding= 3)
```

```python
# I want to create a pie chart to show the contribution of each sub category on Profit

cat_profit = data.groupby('Category')['Sales'].sum().sort_values(ascending = False)

plt.pie(cat_profit, labels = cat_profit.index, autopct = '%1.1f%%')
plt.show()
```
